```python
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score, precision_score, recall_score, roc_auc_score, balanced_accuracy_score
from sklearn.metrics import confusion_matrix
import seaborn as sn
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import roc_curve, auc
import pandas as pd
```

### Scores for Desmedth to TCGA

```python
y_t=r"..\Data\TCGAvsDesmd\desmethtotcga\test\true.csv"
y_true=pd.read_csv(y_t, encoding='utf-8', engine='python')
y_true_labels= y_true
```

#### KPCA-DA

```python
y1=r"..\Data\TCGAvsDesmd\desmethtotcga\test\pred_TCGA.csv"
y_pred1=pd.read_csv(y1, encoding='utf-8', engine='python') 
y_pred_labels1=(y_pred1 > 0.5).astype(int) 
```

```python
accuracy = accuracy_score(y_true_labels, y_pred_labels1)
f1 = f1_score(y_true_labels, y_pred_labels1)
precision = precision_score(y_true_labels, y_pred_labels1)
recall = recall_score(y_true_labels, y_pred_labels1)
specificity = recall_score (y_true_labels, y_pred_labels1,pos_label=0)
bal_acc=balanced_accuracy_score(y_true_labels,y_pred_labels1)
# Print the results
print(f"Accuracy: {accuracy:.2f}")
print(f"F1 Score: {f1:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"Specificity: {specificity:.2f}")
print(f"Bal Acc: {bal_acc:.2f}")
```

#### Combat

```python
y2=r"..\Data\TCGAvsDesmd\desmethtotcga\test\pred_TCGA_Combat.csv"
y_pred2=pd.read_csv(y2, encoding='utf-8', engine='python') 
y_pred_labels2=(y_pred2 > 0.5).astype(int) 
```

```python
fpr1, tpr1, thresholds1 = roc_curve(y_true, y_pred1)
fpr2, tpr2, thresholds2 = roc_curve(y_true, y_pred2)
fpr3, tpr3, thresholds3 = roc_curve(y_true, y_pred3)
fpr4, tpr4, thresholds4 = roc_curve(y_true, y_pred4)


# Compute AUC (Area Under Curve)
roc_auc1 = auc(fpr1, tpr1)
roc_auc2 = auc(fpr2, tpr2)
roc_auc3 = auc(fpr3, tpr3)
roc_auc4 = auc(fpr4, tpr4)



# Plot ROC Curve
plt.figure(figsize=(8, 6))
plt.plot(fpr1, tpr1, label=f'KPC-DA (AUC = {roc_auc1:.2f}±0.03)', color='blue')
plt.plot(fpr2, tpr2, label=f'COMBAT (AUC = {roc_auc2:.2f}±0.03)', color='orange')
plt.plot(fpr3, tpr3, label=f'CAN(AUC = {roc_auc3:.2f}±0.02)', color='green')
plt.plot(fpr4, tpr4, label=f'Without-DA(AUC = {roc_auc4:.2f}±0.04)', color='red')
plt.plot([0, 1], [0, 1], 'k--', label='Random Classifier (AUC = 0.5)')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend(loc='lower right')
plt.show()
plt.savefig("ROC_Curve2")
```
